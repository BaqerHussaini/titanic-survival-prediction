#### __Load the Dataset__

```python
# Import the 'train.csv' dataset from 2-under-process folder
train_df = pd.read_csv('../data/2-under-process/train.csv')
```

```python
# Show the first 5 rows of the dataset
train_df.head()
```

```python
# Get the structural breakdown of the dataset
train_df.info()
```

```python
# Calculate the exact deep RAM footprint of 'train.csv' File in Megabytes (MB)
initial_memory = train_df.memory_usage(deep=True).sum() / (1024**2)
print(f"Initial Memory Footprint: {initial_memory:.2f} MB")
```

```python
# 1. Map your binary dummy features to the smallest possible integer box
int8_columns = ['survival', 'ticket_class', 'sex', 'siblings_spouses', 'parents_children', 'boarding_port_c',
               'boarding_port_q', 'boarding_port_s', 'title_master', 'title_miss', 'title_mr', 'title_mrs',
               'title_rare']
# 2. Map 'passenger_id' column to 16-bit ints
int16_columns = ['passenger_id']

# 3. Map small continuous values to 32-bit floats
float64_columns = ['fare'] 

# 4. Execute the manual type casting on the training data safely
for col in train_df.columns:
    if col in int8_columns:
        train_df[col] = train_df[col].astype('int8')

for col in train_df.columns:
    if col in int16_columns:
        train_df[col] = train_df[col].astype('int16')


for col in train_df.columns:
    if col in float64_columns:
        train_df[col] = train_df[col].astype('float64')

# 4. Calculate the optimized dataset memory usage
final_memory = train_df.memory_usage(deep=True).sum() / (1024**2)
print(f"Optimized Memory Footprint: {final_memory:.2f} MB")
print(f"Total Ram Space Saved: {((initial_memory - final_memory) / initial_memory) * 100:.1f}%")
```

```python
# View the final structural type array
print(train_df.dtypes)
```

```python
# Check if downcasting is done correctly
baseline_df = pd.read_csv('../data/2-under-process/train.csv')

for col in train_df.columns.difference(['age']):
    x =  (train_df[col] == baseline_df[col])
    print("----------------")
    print(f"{x.value_counts()}")
```
