* Sibling = brother, sister, stepbrother, stepsister

__parents_children:__ The dataset defines family relations in this way...
* Parent = mother, father
* Child = daughter, son, stepdaughter, stepson

### __2.1 Working On `train.csv` Dataset__

#### __Stage 1 -->__

#### __Load the Dataset__

```python
# Import the 'train.csv' dataset from 2-under-process folder
train_df = pd.read_csv('../data/2-under-process/train.csv')
```

```python
# Show the first 5 rows of the dataset
train_df.head()
```

```python
# Get the structural breakdown of the dataset
train_df.info()
```

```python
# 1. Create 'title' feature
# 1.1 Extract titles
train_df['title'] = train_df['name'].str.extract(' ([A-Za-z]+)\.', expand=False)

# 1.2 Check unique values and number of missing values
print(f"Number of missing values in 'title' column: {train_df['title'].isnull().sum()}")
train_df['title'].value_counts()
```

```python
# 1.3 Translation and standardize
title_map = {
    'Mlle': 'Miss', 'Mme': 'Mrs', 'Ms': 'Miss',
    'Dr': 'Rare', 'Rev': 'Rare', 'Col': 'Rare', 'Major': 'Rare', 
    'Capt': 'Rare', 'Don': 'Rare', 'Lady': 'Rare', 'Sir': 'Rare', 
    'Countess': 'Rare', 'Jonkheer': 'Rare'
}
train_df['title'] = train_df['title'].replace(title_map)

# 1.4 Check unique values and if the changes applied
train_df['title'].value_counts()
```

#### __Save the Changes__

```python
# Save the data into under-process folder
train_df.to_csv('../data/2-under-process/train.csv', index=False)
```
